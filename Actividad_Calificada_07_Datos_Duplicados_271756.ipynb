## Carga de datos

Ejecuta la siguiente celda. No requiere modificaciones.

```python
import pandas as pd

# Carga del dataset Hotel Booking Demand desde GitHub (TidyTuesday, 2020-02-11)
url = 'https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2020/2020-02-11/hotels.csv'
hoteles = pd.read_csv(url)

# Dimensiones y primeras filas
print('Dimensiones:', hoteles.shape)
hoteles.head()
```

## Actividad 1. Diagnóstico inicial (20 puntos)

### 1.1 Conteo de duplicados exactos (5 pts)

Calcula el número de filas marcadas como duplicadas en `hoteles` considerando **todas** las columnas, con `keep='first'`, `keep='last'` y `keep=False`. Imprime los tres resultados con un texto que explique qué significa cada uno.

```python
# Comparar todas las columnas para identificar filas exactamente iguales.
duplicados_primero = hoteles.duplicated(keep='first').sum()
duplicados_ultimo = hoteles.duplicated(keep='last').sum()
duplicados_todos = hoteles.duplicated(keep=False).sum()

print('Conservar primera: ', duplicados_primero, 'filas sobrantes')
print('Conservar última: ', duplicados_ultimo, 'filas sobrantes')
print('Marcar todas:     ', duplicados_todos, 'filas de grupos repetidos')
```

### 1.2 Magnitud del problema (5 pts)

Con los resultados de 1.1, calcula e imprime:

- El **porcentaje** de filas sobrantes respecto al total.
- El número de **grupos repetidos** (combinaciones de valores que aparecen más de una vez).

```python
# Las filas sobrantes son las que quitaría si conservo una de cada grupo.
porcentaje_sobrantes = duplicados_primero / len(hoteles) * 100
# Cada grupo repetido aporta una fila conservada además de las sobrantes.
grupos_repetidos = duplicados_todos - duplicados_primero

print(f'Filas sobrantes: {porcentaje_sobrantes:.2f}%')
print('Grupos repetidos:', grupos_repetidos)
```

### 1.3 Duplicados por hotel (5 pts)

Determina **cuántas** filas sobrantes (`keep='first'`) hay en cada hotel y qué **porcentaje** representan respecto al total de reservas de ese hotel.

Sugerencia: crea una copia de `hoteles`, agrega una columna booleana `es_duplicado` con el resultado de `duplicated()` y agrupa por `hotel`.

```python
# Marcar duplicados antes de agrupar, para comparar las 32 columnas originales.
revision_hoteles = hoteles.copy()
revision_hoteles['es_duplicado'] = hoteles.duplicated(keep='first')

# La media de los booleanos da la proporción dentro de cada hotel.
duplicados_por_hotel = revision_hoteles.groupby('hotel')['es_duplicado'].agg(['size', 'sum', 'mean'])
duplicados_por_hotel.columns = ['reservas', 'filas_sobrantes', 'proporcion']
duplicados_por_hotel['porcentaje'] = duplicados_por_hotel['proporcion'] * 100
print(duplicados_por_hotel[['reservas', 'filas_sobrantes', 'porcentaje']].to_string())
```

### 1.4 Conclusión (5 pts)

¿Qué tan grave es el problema de duplicados en este dataset? ¿Afecta por igual a ambos hoteles? Responde con cifras.

**Respuesta 1.4:**

Se encontraron 31,994 filas sobrantes de 119,390 (26.80 %) y hay 40,165 filas involucradas en 8,171 grupos repetidos. El porcentaje es considerable para analizarlo, pero no basta para concluir que todas son errores. No afecta igual a los hoteles: City Hotel tiene 25,902 de 79,330 filas sobrantes (32.65 %) y Resort Hotel 6,092 de 40,060 (15.21 %).

## Actividad 2. Inspección y criterio de duplicado (25 puntos)

### 2.1 Ver los grupos repetidos (6 pts)

Filtra las filas involucradas en duplicados (`keep=False`), **ordénalas** con `sort_values()` para que las copias queden juntas y muestra las primeras 12 filas con estas columnas:

`hotel`, `lead_time`, `arrival_date_year`, `arrival_date_month`, `arrival_date_day_of_month`, `adults`, `country`, `market_segment`, `deposit_type`, `adr`, `reservation_status`.

Elige tú las columnas de ordenamiento y explica tu elección en un comentario.

```python
# keep=False incluye también la primera fila de cada grupo repetido.
filas_repetidas = hoteles[hoteles.duplicated(keep=False)]
columnas_vista = ['hotel', 'lead_time', 'arrival_date_year', 'arrival_date_month',
                  'arrival_date_day_of_month', 'adults', 'country', 'market_segment',
                  'deposit_type', 'adr', 'reservation_status']
# Ordenar por hotel, llegada y características de la reserva para juntar casos parecidos.
orden = ['hotel', 'arrival_date_year', 'arrival_date_month',
         'arrival_date_day_of_month', 'lead_time', 'adr', 'reservation_status']
print(filas_repetidas.sort_values(orden)[columnas_vista].head(12).to_string(index=False))
```

### 2.2 Tamaño de los grupos (6 pts)

Usa `value_counts()` sobre el `DataFrame` completo para contar cuántas veces aparece cada combinación de filas. Considera que las columnas `agent` y `company` tienen faltantes (revisa el parámetro `dropna`).

Con ese resultado:

1. Muestra **cuántos grupos** hay de cada tamaño (2 copias, 3 copias, …). Basta con los 10 tamaños más frecuentes.
2. Indica el **tamaño del grupo más grande** y muestra su combinación de valores.

```python
# dropna=False permite incluir también combinaciones con agent o company faltantes.
tamanos_grupo = hoteles.value_counts(dropna=False)
grupos_con_copias = tamanos_grupo[tamanos_grupo > 1]

# Contar cuantos grupos tienen cada cantidad de copias.
frecuencia_tamanos = grupos_con_copias.value_counts()
print('Los 10 tamaños de grupo más frecuentes:')
print(frecuencia_tamanos.head(10).to_string())
print('\nCopias del grupo más grande:', grupos_con_copias.max())
print('Valores de todas sus columnas:')
print(grupos_con_copias.head(1).to_string())
```

### 2.3 ¿En qué se distinguen las filas duplicadas? (6 pts)

Compara las filas involucradas en duplicados (`keep=False`) con el dataset completo en:

- Distribución de `market_segment` (proporciones).
- Distribución de `deposit_type` (proporciones).
- Tasa de cancelación (`is_canceled`).

```python
# Comparar porcentajes del dataset completo contra los grupos repetidos.
segmentos = pd.concat([
    hoteles['market_segment'].value_counts(normalize=True).rename('total'),
    filas_repetidas['market_segment'].value_counts(normalize=True).rename('repetidas')
], axis=1).fillna(0) * 100
depositos = pd.concat([
    hoteles['deposit_type'].value_counts(normalize=True).rename('total'),
    filas_repetidas['deposit_type'].value_counts(normalize=True).rename('repetidas')
], axis=1).fillna(0) * 100

print('Segmento de mercado (%):\n', segmentos.round(2).to_string())
print('\nTipo de depósito (%):\n', depositos.round(2).to_string())
print(f'\nCancelación total: {hoteles["is_canceled"].mean() * 100:.2f}%')
print(f'Cancelación en filas repetidas: {filas_repetidas["is_canceled"].mean() * 100:.2f}%')
```

## Actividad 3. Impacto en el análisis (25 puntos)

### 3.1 Versión sin duplicados (5 pts)

Crea `hoteles_limpio` eliminando los duplicados exactos, conservando la primera aparición y **renumerando el índice**. Imprime el número de filas antes y después y cuántas se eliminaron.

```python
# Quitar solo copias exactas y renumerar el índice para la comparación solicitada.
hoteles_limpio = hoteles.drop_duplicates(keep='first', ignore_index=True)
print('Filas originales:', len(hoteles))
print('Filas después:', len(hoteles_limpio))
print('Filas eliminadas:', len(hoteles) - len(hoteles_limpio))
```

### 3.2 Métricas por hotel (8 pts)

Para `hoteles` y para `hoteles_limpio`, calcula por hotel con `groupby()`:

- Número de reservas (`.size()`).
- Tasa de cancelación (media de `is_canceled`).
- Tarifa diaria promedio (media de `adr`).

Nota: para calcular la media de varias columnas a la vez, selecciona una lista de columnas: `df.groupby('hotel')[['col1', 'col2']].mean()`.

Guarda las medias en dos `DataFrame`, `resumen_original` y `resumen_limpio`, con `hotel` como columna (usa `reset_index()`).

```python
# Calcular medias por hotel y el número de filas con los dos conjuntos de datos.
resumen_original = hoteles.groupby('hotel')[['is_canceled', 'adr']].mean().reset_index()
resumen_original['reservas'] = hoteles.groupby('hotel').size().values
resumen_limpio = hoteles_limpio.groupby('hotel')[['is_canceled', 'adr']].mean().reset_index()
resumen_limpio['reservas'] = hoteles_limpio.groupby('hotel').size().values

# Poner nombres fáciles de interpretar en la tabla final.
resumen_original = resumen_original.rename(columns={'is_canceled': 'tasa_cancelacion', 'adr': 'tarifa_promedio'})
resumen_limpio = resumen_limpio.rename(columns={'is_canceled': 'tasa_cancelacion', 'adr': 'tarifa_promedio'})
print('Original:\n', resumen_original.to_string(index=False))
print('\nSin duplicados:\n', resumen_limpio.to_string(index=False))
```

### 3.3 Tabla comparativa (6 pts)

Agrega a cada resumen una columna `version` (`'original'` y `'sin duplicados'`) y únelos con `pd.concat()` en un `DataFrame` llamado `comparacion`, con índice renumerado y ordenado por `hotel`, de modo que las dos versiones de cada hotel queden juntas.

```python
# Etiquetar cada versión antes de unirlas y ordenar ambos hoteles.
resumen_original['version'] = 'original'
resumen_limpio['version'] = 'sin duplicados'
comparacion = pd.concat([resumen_original, resumen_limpio], ignore_index=True)
comparacion = comparacion.sort_values(['hotel', 'version']).reset_index(drop=True)
print(comparacion.to_string(index=False))
```
