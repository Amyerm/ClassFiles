## Carga de datos

Ejecuta la siguiente celda. No requiere modificaciones.

```python
import pandas as pd

# Carga del dataset de taxis de Nueva York desde un repositorio público de GitHub.
# El archivo se lee con pandas; no se requiere ninguna otra biblioteca.
url = 'https://raw.githubusercontent.com/mwaskom/seaborn-data/master/taxis.csv'
taxis = pd.read_csv(url)

print('Dimensiones:', taxis.shape)
taxis.head()
```

## Actividad 1. Conversión y validación (20 puntos)

### 1.1 Conversión (5 pts)

1. Muestra el tipo de dato de `pickup` y `dropoff`.
2. Convierte ambas columnas a tipo fecha con `pd.to_datetime()`, indicando el **formato explícito** que corresponde a los valores.
3. Verifica que la conversión no generó `NaT`: compara el número de faltantes antes y después.

```python
# Primero revisar como se leyeron las dos columnas de fechas.
print('Tipos antes de convertir:')
print(taxis[['pickup', 'dropoff']].dtypes)
faltantes_antes = taxis[['pickup', 'dropoff']].isna().sum()

# Indicar el formato de año-mes-día y hora:minuto:segundo. Si una fecha
# no se pudiera convertir, errors='coerce' hace verla como NaT.
for columna in ['pickup', 'dropoff']:
    taxis[columna] = pd.to_datetime(taxis[columna], format='%Y-%m-%d %H:%M:%S', errors='coerce')

print('\nTipos después de convertir:')
print(taxis[['pickup', 'dropoff']].dtypes)
print('\nFaltantes antes y después:')
print(pd.DataFrame({'antes': faltantes_antes,
                    'después': taxis[['pickup', 'dropoff']].isna().sum()}))
```

### 1.2 Periodo de los datos (5 pts)

El dataset debería contener solo viajes de marzo de 2019.

1. Muestra la fecha mínima y máxima de `pickup` y de `dropoff`.
2. Cuenta y muestra los viajes cuyo `pickup` **no** está en marzo de 2019.
3. ¿Hay viajes con `pickup` en marzo y `dropoff` fuera de marzo? Cuéntalos.

```python
# Uso el inicio del 1 de abril como límite superior sin incluirlo.
inicio_marzo = pd.Timestamp('2019-03-01')
inicio_abril = pd.Timestamp('2019-04-01')
pickup_en_marzo = taxis['pickup'].between(inicio_marzo, inicio_abril, inclusive='left')
dropoff_en_marzo = taxis['dropoff'].between(inicio_marzo, inicio_abril, inclusive='left')

print('Rango de pickup:', taxis['pickup'].min(), 'a', taxis['pickup'].max())
print('Rango de dropoff:', taxis['dropoff'].min(), 'a', taxis['dropoff'].max())
print('\nViajes que iniciaron fuera de marzo:', (~pickup_en_marzo).sum())
print(taxis.loc[~pickup_en_marzo, ['pickup', 'dropoff']].to_string(index=False))
print('\nIniciaron en marzo y terminaron fuera:',
      (pickup_en_marzo & ~dropoff_en_marzo).sum())
print(taxis.loc[pickup_en_marzo & ~dropoff_en_marzo,
                ['pickup', 'dropoff']].to_string(index=False))
```

### 1.3 Cambio de horario (5 pts)

```python
# Revisar las horas de inicio del domingo en que cambió el reloj.
dia_cambio = pd.Timestamp('2019-03-10').date()
viajes_cambio = taxis.loc[taxis['pickup'].dt.date == dia_cambio]
conteo_horas = viajes_cambio.groupby(viajes_cambio['pickup'].dt.hour).size()
print('Viajes que iniciaron el 10 de marzo:', len(viajes_cambio))
print('\nViajes por hora:')
print(conteo_horas.to_string())
print('Viajes iniciados entre las 02:00 y las 02:59:', conteo_horas.get(2, 0))

# Marcar como NaT las horas que no existieron en Nueva York.
pickup_ny = taxis['pickup'].dt.tz_localize('America/New_York',
                                         nonexistent='NaT', ambiguous='NaT')
print('Valores NaT al localizar pickup:', pickup_ny.isna().sum())
```

## Actividad 2. Duraciones y anomalías (25 puntos)

### 2.1 Duración de cada viaje (6 pts)

1. Crea la columna `duracion` = `dropoff` − `pickup`.
2. Crea `duracion_min` con la duración en **minutos** (usa `.dt.total_seconds()`).
3. Muestra el resumen estadístico de `duracion_min` con `describe()`.

```python
# La resta de dos fechas produce una duración, convertir sus segundos a minutos.
taxis['duracion'] = taxis['dropoff'] - taxis['pickup']
taxis['duracion_min'] = taxis['duracion'].dt.total_seconds() / 60
print(taxis['duracion_min'].describe())
```

### 2.2 Duraciones anómalas (6 pts)

Cuenta los viajes con:

- Duración **negativa** (termina antes de iniciar).
- Duración **cero**.
- Duración **menor a 1 minuto** (incluye los de duración cero).
- Duración **mayor a 3 horas**.

Muestra los viajes de duración cero con las columnas `pickup`, `dropoff`, `distance`, `fare` y `total`.

```python
# Cada máscara busca un tipo de duración sospechosa.
duracion_negativa = taxis['duracion_min'] < 0
duracion_cero = taxis['duracion_min'] == 0
duracion_corta = taxis['duracion_min'] < 1
duracion_larga = taxis['duracion_min'] > 180

print('Duración negativa:', duracion_negativa.sum())
print('Duración cero:', duracion_cero.sum())
print('Menor a un minuto (incluye las de cero):', duracion_corta.sum())
print('Mayor a tres horas:', duracion_larga.sum())
print('\nViajes con duración cero:')
print(taxis.loc[duracion_cero,
                ['pickup', 'dropoff', 'distance', 'fare', 'total']].to_string(index=False))

# Mirar distancia y tarifa para decidir qué significan los viajes cortos.
print('\nViajes cortos con distancia positiva:',
      (duracion_corta & (taxis['distance'] > 0)).sum())
print('Viajes cortos con tarifa positiva:',
      (duracion_corta & (taxis['fare'] > 0)).sum())
```

### 2.3 Velocidad promedio (6 pts)

La velocidad promedio de un viaje, en millas por hora, es `distance / (duracion_min / 60)`.

1. Crea la columna `velocidad_mph` **solo para viajes con duración mayor que cero**; en los demás debe quedar como faltante (`NaN`). Sugerencia: calcula la velocidad para todos los viajes y conserva solo los valores válidos con `.where(condición)`, que deja `NaN` donde la condición es `False`.
2. Cuenta los viajes con velocidad mayor a 60 mph (velocidad poco plausible en tráfico urbano).
3. Muestra los 5 viajes con mayor velocidad, ordenados de mayor a menor, con `pickup`, `dropoff`, `duracion_min`, `distance`, `velocidad_mph` y `fare`.

```python
# Dividir las millas entre el tiempo en horas. Los viajes de duración
# cero no tienen una velocidad calculable, así que dejo NaN en ellos.
taxis['velocidad_mph'] = (taxis['distance'] /
                          (taxis['duracion_min'] / 60)).where(taxis['duracion_min'] > 0)
velocidad_alta = taxis['velocidad_mph'] > 60
print('Viajes con velocidad mayor a 60 mph:', velocidad_alta.sum())
print('De ellos, con duración menor a un minuto:',
      (velocidad_alta & duracion_corta).sum())
print('\nCinco mayores velocidades:')
columnas = ['pickup', 'dropoff', 'duracion_min', 'distance', 'velocidad_mph', 'fare']
print(taxis.sort_values('velocidad_mph', ascending=False)[columnas]
      .head(5).to_string(index=False))
```

## Actividad 3. Componentes y patrones (25 puntos)

### 3.1 Extraer componentes (6 pts)

Crea las columnas:

- `hora`: hora de `pickup`.
- `dia_num`: día de la semana de `pickup` (lunes = 0).
- `dia_nombre`: nombre del día de `pickup`.
- `fin_semana`: `True` si es sábado o domingo.
- `fecha`: solo la fecha (sin hora) de `pickup`.
- `cruza_medianoche`: `True` si la **fecha** de `dropoff` es distinta de la de `pickup`.

Cuenta cuántos viajes cruzan la medianoche.

```python
# Extraer las partes de la fecha de inicio para agrupar los viajes.
taxis['hora'] = taxis['pickup'].dt.hour
taxis['dia_num'] = taxis['pickup'].dt.dayofweek
taxis['dia_nombre'] = taxis['pickup'].dt.day_name()
taxis['fin_semana'] = taxis['dia_num'] >= 5
taxis['fecha'] = taxis['pickup'].dt.date

# Comparar solo las fechas para detectar viajes que terminan otro día.
taxis['cruza_medianoche'] = taxis['pickup'].dt.date != taxis['dropoff'].dt.date
print('Viajes que cruzan la medianoche:', taxis['cruza_medianoche'].sum())
print(taxis[['pickup', 'dropoff', 'hora', 'dia_nombre',
             'fin_semana', 'fecha', 'cruza_medianoche']].head())
```

### 3.2 Patrón por hora del día (7 pts)

Con `groupby('hora')`, calcula para cada hora: el **número de viajes** y la **mediana** de `duracion_min`. Indica la hora con más viajes y la hora con menos.

Usa la mediana (no la media) porque la duración tiene valores extremos.

```python
# La mediana reduce el efecto de los viajes de duración extrema.
por_hora = taxis.groupby('hora').agg(
    viajes=('hora', 'size'),
    duracion_mediana=('duracion_min', 'median')
)
print(por_hora)
print('\nHora con más viajes:', por_hora['viajes'].idxmax(),
      'con', por_hora['viajes'].max())
print('Hora con menos viajes:', por_hora['viajes'].idxmin(),
      'con', por_hora['viajes'].min())
print('Hora con mayor duración mediana:', por_hora['duracion_mediana'].idxmax(),
      'con', round(por_hora['duracion_mediana'].max(), 2), 'minutos')
```

### 3.3 Patrón por día de la semana (6 pts)

Calcula por día de la semana el **número de viajes** y la **mediana** de `duracion_min` y de `fare`. Presenta los días en orden de lunes a domingo (sugerencia: agrupa por `['dia_num', 'dia_nombre']`).

Toma en cuenta que marzo de 2019 no tiene el mismo número de cada día de la semana: calcula también el **promedio de viajes por fecha**, dividiendo entre el número de fechas distintas de ese día de la semana (usa la columna `fecha` y `.nunique()`).

```python
# Agrupar por número y nombre para mantener el orden lunes a domingo.
por_dia = taxis.groupby(['dia_num', 'dia_nombre']).agg(
    viajes=('fecha', 'size'),
    duracion_mediana=('duracion_min', 'median'),
    tarifa_mediana=('fare', 'median'),
    fechas_distintas=('fecha', 'nunique')
).sort_index()

# Dividir entre las fechas presentes de cada día, no entre siete.
por_dia['viajes_por_fecha'] = por_dia['viajes'] / por_dia['fechas_distintas']
print(por_dia.round(2).to_string())
```

## Actividad 4. Dataset limpio y comparación (20 puntos)

### 4.1 Dataset limpio (6 pts)

Crea `taxis_limpio` aplicando el criterio que definiste en 2.4. Imprime cuántos viajes se excluyeron por cada condición y en total, y verifica que `taxis_limpio` cumple todas las condiciones.

```python
# Al aplicar limites de 2.4, una fila puede cumplir más
# de una condición, por eso el total se obtiene con una sola máscara OR.
fuera_periodo = ~pickup_en_marzo
corta_o_negativa = taxis['duracion_min'] < 1
muy_larga = taxis['duracion_min'] > 180
muy_rapida = taxis['velocidad_mph'] > 60
excluir = fuera_periodo | corta_o_negativa | muy_larga | muy_rapida

taxis_limpio = taxis.loc[~excluir].copy()
print('Fuera de marzo:', fuera_periodo.sum())
print('Duración menor a 1 minuto (incluye cero y negativas):', corta_o_negativa.sum())
print('Duración mayor a 180 minutos:', muy_larga.sum())
print('Velocidad mayor a 60 mph:', muy_rapida.sum())
print('Viajes excluidos sin contarlos dos veces:', excluir.sum())
print('Antes:', len(taxis), '| Después:', len(taxis_limpio))

# Verificar que el resultado sí cumple el criterio.
print('Fuera de marzo en el dataset limpio:',
      (~taxis_limpio['pickup'].between(inicio_marzo, inicio_abril, inclusive='left')).sum())
print('Duraciones fuera del intervalo [1, 180] minutos:',
      (~taxis_limpio['duracion_min'].between(1, 180)).sum())
print('Velocidades mayores a 60 mph:',
      (taxis_limpio['velocidad_mph'] > 60).sum())
print('Velocidades faltantes:', taxis_limpio['velocidad_mph'].isna().sum())
```

### 4.2 Comparación entre periodos (6 pts)

Con `taxis_limpio`, compara los viajes que inician entre las **7:00 y las 9:59** en **días hábiles** contra los del **mismo horario en fin de semana**. Para cada grupo reporta: número de viajes, mediana de `duracion_min` y mediana de `velocidad_mph`.

```python
# Comparar los inicios de 07:00 a 09:59, separando días hábiles y fin de semana.
por_la_manana = taxis_limpio['hora'].between(7, 9)
comparacion = taxis_limpio.loc[por_la_manana].groupby('fin_semana').agg(
    viajes=('hora', 'size'),
    duracion_mediana_min=('duracion_min', 'median'),
    velocidad_mediana_mph=('velocidad_mph', 'median'),
    fechas_distintas=('fecha', 'nunique')
)
comparacion['viajes_por_fecha'] = comparacion['viajes'] / comparacion['fechas_distintas']
comparacion.index = comparacion.index.map({False: 'Días hábiles', True: 'Fin de semana'})
print(comparacion.round(2).to_string())
```
